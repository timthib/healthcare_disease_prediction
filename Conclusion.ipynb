# CONCLUSION


Ce projet avait pour objectif d’analyser les déterminants de la mortalité infantile à travers le monde, en dépassant le simple constat de la croissance économique pour explorer le rôle des infrastructures et de l’efficacité des systèmes de santé.


Nous avons tout d’abord procédé à une collecte automatisée et à un nettoyage ciblé, excluant les variables trop lacunaires comme l’indice de Gini. Une étape cruciale a été la réduction dimensionnelle par Analyse en Composantes Principales (ACP). Nous avons ainsi pu transformer une dizaine de variables corrélées en trois scores synthétiques : la richesse/infrastructure, l’efficacité structurelle et la démographie. Cette approche a permis de résoudre les problèmes de multicolinéarité qui auraient biaisé nos estimations ultérieures.


L’exploration visuelle de nos données a permis de confirmer que si le PIB est un socle indispensable, les ressources humaines (densité de médecins) et l’efficacité de gestion des soins créent des divergences majeures entre pays de richesse équivalente. 

L’étude des trajectoires de développement et des scores de performance a révélé deux enseignements majeurs :

•	L’efficacité relative : Des pays comme la Thaïlande ou l'Ouganda affichent des résultats sanitaires bien supérieurs à ce que leur niveau de richesse laisserait prédire, prouvant qu'une organisation efficace des soins et une forte densité de personnel soignant (médecins et infirmiers) peuvent compenser un revenu modeste.

•	Le rendement marginal des investissements : L’analyse logarithmique a démontré que les investissements dans le personnel médical ont un impact proportionnellement beaucoup plus massif dans les pays à bas revenus. Passer de 1 à 5 médecins pour 10 000 habitants sauve bien plus de vies que d’ajouter des médecins dans un système déjà performant.


Dès lors, pour valider statistiquement ces observations et répondre à notre problématique initiale, à savoir si l’organisation de l’offre de soins prime sur la richesse brute, il est devenu nécessaire de passer d’une analyse descriptive à une approche de modélisation statistique. Cette étape nous a permis de quantifier précisément le poids de chaque levier tout en tenant compte de la dynamique temporelle et des spécificités historiques de chaque pays.


Nous avons comparé deux méthodes : une approche statistique classique (la régression linéaire) et une technique d'intelligence artificielle plus complexe (XGBoost).

•	Le modèle de régression avec lag (t-1) s’est révélé être le plus précis pour la prédiction à court terme (RMSE de 0,55 sur le test), soulignant que la mortalité infantile est un processus doté d’une forte inertie temporelle.

•	Le modèle XGBoost, bien que capturant des relations non linéaires complexes, a montré des limites sur notre échantillon réduit (RMSE de 1,28 sur le test). Sa tendance à sous-estimer les valeurs de mortalité très élevées met en lumière la difficulté de prédire des événements extrêmes lorsque ceux-ci sont sous-représentés dans le jeu de données.


En conclusion, la richesse est un facteur nécessaire, mais le personnel de soins de proximité est le véritable moteur de la survie infantile.

Toutefois, ce projet rencontre des limites qui ouvrent la voie à des recherches futures. L’exclusion de la variable vaccination DTP3, due à un biais de survie statistique, rappelle que les indicateurs administratifs doivent être manipulés avec précaution. De plus, nos modèles gagneraient en fiabilité avec un échantillon plus large de pays à bas revenus pour mieux apprendre les dynamiques de crises. Enfin, une ouverture vers des variables environnementales (accès à l’eau potable, pollution) ou politiques (stabilité institutionnelle) permettrait de compléter cette vision et d’affiner les recommandations en matière de politiques de santé publique.
